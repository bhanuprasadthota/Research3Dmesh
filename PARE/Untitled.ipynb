```python
import os
from collections import defaultdict

def get_folder_size(path):
    """Calculate the total size of a folder."""
    total_size = 0
    for dirpath, _, filenames in os.walk(path):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            if os.path.isfile(fp):
                total_size += os.path.getsize(fp)
    return total_size

def find_duplicate_folders(directory):
    """Find duplicate folders based on name and size."""
    folder_data = defaultdict(list)

    for item in os.listdir(directory):
        item_path = os.path.join(directory, item)
        if os.path.isdir(item_path):
            folder_size = get_folder_size(item_path)
            # Store folder path in a dictionary with key as (folder_name, size)
            folder_data[(item, folder_size)].append(item_path)

    # Identify duplicates
    duplicates = {k: v for k, v in folder_data.items() if len(v) > 1}

    if duplicates:
        print(f"Found {len(duplicates)} duplicate folders:\n")
        for (name, size), paths in duplicates.items():
            print(f"Folder Name: {name}, Size: {size} bytes")
            print("Paths:")
            for path in paths:
                print(f"  - {path}")
            print()
    else:
        print("No duplicate folders found.")

# Define the target directory
output_dir = "/x4/ljiao/Bhanu/PARE/output"
find_duplicate_folders(output_dir)
```

```python
import os

def check_folder_content(folder_path):
    """Check if the folder contains exactly 6 items with the specified structure."""
    # List all items in the folder
    items = os.listdir(folder_path)
    item_count = len(items)

    # Counters
    folder_count = 0
    mp4_count = 0
    log_count = 0
    pkl_count = 0

    for item in items:
        item_path = os.path.join(folder_path, item)
        
        if os.path.isdir(item_path):
            folder_count += 1
        elif os.path.isfile(item_path):
            if item.endswith(".mp4"):
                mp4_count += 1
            elif item.endswith(".log"):
                log_count += 1
            elif item.endswith(".pkl"):
                pkl_count += 1

    # Check for the expected structure
    if item_count != 6 or folder_count != 2 or mp4_count != 2 or log_count != 1 or pkl_count != 1:
        print(f"❌ {folder_path} - Incorrect structure")
        print(f"  - Folders: {folder_count}")
        print(f"  - .mp4 Files: {mp4_count}")
        print(f"  - .log Files: {log_count}")
        print(f"  - .pkl Files: {pkl_count}")
        print(f"  - Total Items: {item_count}\n")

def find_target_folders(base_path):
    """Find folders ending with `_` and check their structure."""
    for root, subdirs, _ in os.walk(base_path):
        for subdir in subdirs:
            # Check for folders ending with `_`
            if subdir.endswith("_"):
                folder_path = os.path.join(root, subdir)
                check_folder_content(folder_path)

# Define the base directory
base_dir = "/x4/ljiao/Bhanu/PARE/output"
find_target_folders(base_dir)
```
